```python
def recommend_item(city, time, item, cart_size):

    if item == "pizza":
        return "coke"

    if item == "burger":
        return "fries"

    if item == "biryani":
        return "raita"

    if time == "morning":
        return "coffee"

    if cart_size >= 3:
        return "dessert"

    return "ice cream"
```

```python
df["model_recommendation"] = df.apply(
    lambda x: recommend_item(
        x["city"],
        x["order_time"],
        x["item_ordered"],
        x["cart_size"]
    ),
    axis=1
)

df.head()
```
