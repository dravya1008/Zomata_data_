```python
import pandas as pd
df = pd.read_csv("popular_food_combos_dataset.csv")
df.head()
```

```python
popular_combos = df.sort_values(by="popularity_score", ascending=False)
popular_combos.head(5)
```

```python
df.groupby("city")["combo"].value_counts().head(10)
```

```python
def recommend_for_new_user(city):
    
    city_data = df[df["city"] == city]
    
    popular = city_data.sort_values(
        by="popularity_score",
        ascending=False
    )
    
    return popular["combo"].head(3)
```

```python
recommend_for_new_user("Delhi")
```
